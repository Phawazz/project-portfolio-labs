```python
# Identify and handle duplicates across the entire dataset
print(df_hour.duplicated().sum())
```

```python
# Drop the 'casual' and 'registered' columns since they both add up to form the 'count' column
df_hour = df_hour.drop(['casual', 'registered'], axis=1)
```

```python
# Convert the datetime from object to a datetime dtype first
df_hour['datetime'] = pd.to_datetime(df_hour['datetime'])

# Extract the datetime components into separate columns
df_hour['hour'] = df_hour['datetime'].dt.hour
df_hour['month'] = df_hour['datetime'].dt.month
df_hour['dayofweek'] = df_hour['datetime'].dt.dayofweek
df_hour['year'] = df_hour['datetime'].dt.year

# Drop the object column
df_hour.drop('datetime', axis=1, inplace=True)
```

```python
# Review the data again
df_hour.head()
```

### Data Formatting
Ensuring Appropriate data types for the sake of EDA

```python
# Map the seasons based on kaggle documentation
season_map = {
    1: 'Spring',
    2: 'Summer',
    3: 'Fall',
    4: 'Winter'
}

df_hour['season'] = df_hour['season'].map(season_map)
df_hour['season'].head()
```

```python
# Map weather for clearer charts (based on docs too)
weather_map = {
    1: 'Clear',
    2: 'Misty',
    3: 'Light_Precip',
    4: 'Heavy_Precip'
}

df_hour['weather'] = df_hour['weather'].map(weather_map)
df_hour['weather'].head()
```

```python
# Change data type from int to categorical for the variable 'month'
month_string = {
    1: 'January',
    2: 'February',
    3: 'March',
    4: 'April',
    5: 'May',
    6: 'June',
    7: 'July',
    8: 'August',
    9: 'September',
    10: 'October',
    11: 'November',
    12: 'December'
}

df_hour['month'] = df_hour['month'].map(month_string)
df_hour['month'].head()
```

```python
# Change data type from int to categorical for the variable 'dayofweek'
weekday_string = {
    1: 'Monday',
    2: 'Tuesday',
    3: 'Wednesday',
    4: 'Thursday',
    5: 'Friday',
    6: 'Saturday',
    '7': 'Sunday'
}

df_hour['dayofweek'] = df_hour['dayofweek'].map(weekday_string)
df_hour['dayofweek'].head()
```

### Data Normalisation
To ensure consistency and comparability & accoring to docs provided by UCI ML repo

```python
# Normalise data
df_hour['temp'] = (df_hour['temp']) * 47 - 8
df_hour['atemp'] = (df_hour['atemp']) * 66 - 16
df_hour['humidity'] = (df_hour['humidity']) * 100
df_hour['windspeed'] = (df_hour['windspeed']) * 67
```
